```python

def is_variable(x):
    return isinstance(x, str) and x.islower() and len(x) == 1


def occurs_check(var, expr, subst):
    expr = apply_substitution(expr, subst)

    if var == expr:
        return True

    if isinstance(expr, tuple):
        return any(occurs_check(var, item, subst) for item in expr[1:])

    return False


def apply_substitution(expr, subst):
    if is_variable(expr) and expr in subst:
        return apply_substitution(subst[expr], subst)

    if isinstance(expr, tuple):
        return (expr[0],) + tuple(
            apply_substitution(item, subst) for item in expr[1:]
        )

    return expr


def unify(x, y, subst):
    x = apply_substitution(x, subst)
    y = apply_substitution(y, subst)

    # Step 1: If both expressions are identical
    if x == y:
        return subst

    # Step 2: If first expression is a variable
    if is_variable(x):
        if occurs_check(x, y, subst):
            return None

        subst[x] = y
        return subst

    # Step 3: If second expression is a variable
    if is_variable(y):
        if occurs_check(y, x, subst):
            return None

        subst[y] = x
        return subst

    # Step 4: Check predicate/function symbols
    if isinstance(x, tuple) and isinstance(y, tuple):

        if x[0] != y[0]:
            return None

        # Step 5: Check number of arguments
        if len(x) != len(y):
            return None

        # Step 6: Unify arguments one by one
        for i in range(1, len(x)):
            result = unify(x[i], y[i], subst)

            if result is None:
                return None

        return subst

    # Different constants or incompatible expressions
    return None


# Main program
print("FIRST ORDER LOGIC - UNIFICATION")

# Example: Eats(x, Apple) and Eats(Riya, y)
expr1 = ("Eats", "x", "Apple")
expr2 = ("Eats", "Riya", "y")

result = unify(expr1, expr2, {})

if result is None:
    print("Unification Failed!")
elif result == {}:
    print("NIL - No substitution required")
else:
    print("Unification Successful!")
    print("Substitution:", result)

    print("Unified Expression 1:",
          apply_substitution(expr1, result))

    print("Unified Expression 2:",
          apply_substitution(expr2, result))
```
